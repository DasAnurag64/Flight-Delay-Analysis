```python
import pandas as pd
import numpy as np
Airline=pd.read_csv("D:\Cricket score Analysis\Airline delay analysis\DimAirline_raw.csv")
Originairport=pd.read_csv("D:\Cricket score Analysis\Airline delay analysis\DimAirport_raw.csv")
Destinationairport=pd.read_csv("D:\Cricket score Analysis\Airline delay analysis\DimAirport_raw.csv")
Factflight=pd.read_csv("D:\Cricket score Analysis\Airline delay analysis\FactFlight_raw.csv")
Factflight.info()
Airline.AirlineKey.duplicated()
Originairport.AirportKey.duplicated()
dup=Factflight.FlightKey.duplicated()
Factflight.FlightKey.value_counts().sort_values(ascending=False)
Factflight.drop_duplicates(subset=['FlightKey'], keep='first', inplace=True)
Factflight.info()
Factflight.head(20)
Factflight.describe()
Factflight.sort_values(by='DepartureDelayMinutes',ascending=False)
glitch_indices=Factflight[Factflight.DepartureDelayMinutes==9999].index
Factflight.drop(index=glitch_indices,inplace=True)
Total_canceled_flight=Factflight.query("Cancelled==1")["FlightKey"].count()
Factflight.info()
Factflight.loc[Factflight['Cancelled']==1,'DepartureDelayMinutes']=np.nan
Factflight.loc[Factflight['Cancelled']==1,'ArrivalDelayMinutes']=np.nan
Factflight.query("Cancelled==1")[["DepartureDelayMinutes","ArrivalDelayMinutes"]]
Factflight.describe()
Factflight.DistanceMiles=Factflight.DistanceMiles.abs()
Factflight.describe()
Factflight.head(20)
Factflight["DateKey"]=pd.to_datetime(Factflight["DateKey"].astype("string"),format='%Y%m%d')
Factflight.info()
Factflight.head()

Factflight.loc[Factflight["Cancelled"]==0,"ArrivalDelayMinutes"].fillna(0)
Factflight.loc[Factflight["Cancelled"]==0,"DepartureDelayMinutes"].fillna(0)
Factflight.info()
Factflight.head()



```

```python
avg_Departdelayedtime=Factflight.DepartureDelayMinutes.mean()
avg_arrivaldelayedtime=Factflight.ArrivalDelayMinutes.mean()
cancelled_flight=Factflight.query("Cancelled==1")['Cancelled'].sum()
print(f"Total number of flight cancelled is:{cancelled_flight}")
Diverted_flight=Factflight.query("Diverted==1")['Diverted'].sum()
print(f"Total number of flights diverted is:{Diverted_flight}")
completed=len(Factflight)-cancelled_flight
print(f"Totao No. of completed flight is:{completed}")
departdelay_count=Factflight.query("DepartureDelayMinutes > 15")['FlightKey'].count()
print(f"Count of dparture delayed flights:{departdelay_count}")
arrivaldelay_count=Factflight.query("ArrivalDelayMinutes>15")['FlightKey'].count()
print(f"Count of arrival delayed flights:{arrivaldelay_count}")
cancel_rate=(cancelled_flight/len(Factflight) * 100).round()
print(f"flight cancelation rate:{cancel_rate}")
departdelay_rate=(departdelay_count/completed * 100).__round__(2)
print(f"The departure delay rate is:{departdelay_rate}")
arrivaldelay_rate=(arrivaldelay_count/completed * 100).round()
print(f"The arrival delay rate is:{arrivaldelay_rate}")

```
